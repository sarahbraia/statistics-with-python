```python
# Qui Quadrado
```

```python
# Importação das funções, chi2_contingency porque temos 2 categorias
import numpy as np
from scipy.stats import chi2_contingency
```

```python
# Criação da matriz com os dados e execução do teste
# Criando a matriz de contingência (linhas = sexo, colunas = assiste/não assiste)
# Linha 1: Masculino (19 assiste, 6 não assiste)
# Linha 2: Feminino (43 assiste, 32 não assiste)
novela = np.array([[19, 6], [43, 32]])
novela
```

Teste do Qui-Quadrado de Independência:

Hipótese nula (H₀): Não há associação entre o gênero e o fato de assistir.

Hipótese alternativa (H₁): Existe associação entre o gênero e o fato de assistir.

```python
# Aplicando o teste do Qui-Quadrado
chi2, p, dof, expected = chi2_contingency(novela)

# Exibindo os resultados
print(f"Estatística Qui-Quadrado: {chi2:.4f}")
print(f"Valor-p: {p:.4f}")
print(f"Graus de liberdade: {dof}")
print("Frequências esperadas:")
print(expected)
```

```python
# Criação da matriz com os dados e execução do teste
# Criando a matriz de contingência (linhas = sexo, colunas = assiste/não assiste)
# Linha 1: Masculino (22 assiste, 3 não assiste)
# Linha 2: Feminino (43 assiste, 32 não assiste)
novela2 = np.array([[22, 3], [43, 32]])
novela2
```

```python
# Aplicando o teste do Qui-Quadrado
chi2, p, dof, expected = chi2_contingency(novela2)

# Exibindo os resultados
print(f"Estatística Qui-Quadrado: {chi2:.4f}")
print(f"Valor-p: {p:.4f}")
print(f"Graus de liberdade: {dof}")
print("Frequências esperadas:")
print(expected)
```
