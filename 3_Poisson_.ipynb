```python
# Distribuição de Poisson
```

```python
# Importação da função
from scipy.stats import poisson
```

#### Média de acidentes de carro é 2 por dia

##### Qual a probabilidade de ocorrerem 3 acidentes no dia?

```python
# Parâmetros
lambda_media = 2  # média de acidentes por dia
k = 3             # acidentes desejados

# Cálculo da probabilidade
prob = poisson.pmf(k, lambda_media)

print(f'Probabilidade de ocorrerem exatamente {k} acidentes: {prob:.4f} ou {prob*100:.2f}%')
```

##### Qual a probabilidade de ocorrerem 3 ou menos acidentes no dia?

```python
# Cálculo da probabilidade acumulada
prob_acumulada = poisson.cdf(k, lambda_media)

print(f'Probabilidade de ocorrerem até {k} acidentes: {prob_acumulada:.4f} ou {prob_acumulada*100:.2f}%')
```

##### Qual a probabilidade de ocorrerem mais de 3 acidentes no dia?

```python
# Probabilidade acumulada até 3
p_menor_igual_3 = poisson.cdf(k, lambda_media)

# Probabilidade de mais de 3
p_maior_que_3 = 1 - p_menor_igual_3

print(f'Probabilidade de mais de {k} acidentes: {p_maior_que_3:.4f} ou {p_maior_que_3*100:.2f}%')
```
