```python
# Importação da função binom
from scipy.stats import binom
```

#### Jogar uma moeda 5 vezes, qual a probabilidade de dar cara 3 vezes?
##### eventos , experimentos, probabilidades

```python
# Jogar uma moeda 5 vezes, qual a probabilidade de dar cara 3 vezes?
# eventos , experimentos, probabilidades

# Parâmetros
n = 5        # número de lançamentos
k = 3        # número de caras desejadas
p = 0.5      # probabilidade de sair cara

# Cálculo da probabilidade
probabilidade = binom.pmf(k, n, p)

print(f'Probabilidade de sair exatamente {k} caras em {n} lançamentos: {probabilidade:.4f} ou {probabilidade*100:.2f}%')
```

#### Passar por 4 sinais de 4 tempos, qual a probabilidade de pegar sinal verde
##### nenhuma, 1, 2, 3 ou 4 vezes seguidas?

```python
# Parâmetros
n = 4        # número de sinais
p = 0.25     # probabilidade de pegar sinal verde

# Calcular probabilidade para 0 a 4 sinais verdes
for k in range(5):
    prob = binom.pmf(k, n, p)
    print(f'Probabilidade de pegar {k} sinais verdes: {prob:.4f} ou {prob*100:.2f}%')
```

#### E se forem sinais de dois tempos?

```python
# Parâmetros
n = 4       # número de sinais
p = 0.5     # probabilidade de pegar sinal verde

# Calcular probabilidade para 0 a 4 sinais verdes
for k in range(5):
    prob = binom.pmf(k, n, p)
    print(f'Probabilidade de pegar {k} sinais verdes: {prob:.4f} ou {prob*100:.2f}%')
```

#### Probabilidade acumulativa

```python
# Parâmetros
n = 4        # número de sinais
p = 0.5      # probabilidade de sinal estar verde

# Calcular probabilidade acumulada para cada valor de k (0 a 4)
for k in range(5):
    prob_acumulada = binom.cdf(k, n, p)
    print(f'Pegar até {k} sinais verdes: {prob_acumulada:.4f} ou {prob_acumulada*100:.2f}%')
```

#### Concurso com 12 questões, qual a probabilidade de acertar 7 questões considerando que cada questão tem 4 alternativas?

```python
# Parâmetros
n = 12      # número de questões
k = 7       # número de acertos desejados
p = 0.25    # chance de acertar cada questão

# Probabilidade de acertar exatamente 7 questões
prob = binom.pmf(k, n, p)

print(f'Probabilidade de acertar exatamente {k} questões: {prob:.6f} ou {prob*100:.4f}%')
```

#### Probabilidade de acertar as 12 questões

```python
# Parâmetros
n = 12      # número de questões
k = 12      # número de acertos desejados
p = 0.25    # chance de acertar cada questão

# Probabilidade de acertar exatamente 7 questões
prob = binom.pmf(k, n, p)

print(f'Probabilidade de acertar exatamente {k} questões: {prob:.10f} ou {prob*100:.8f}%')
```
